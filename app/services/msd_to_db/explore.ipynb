```python
import pandas as pd
from pathlib import Path

# ── File path ────────────────────────────────────────────────────────────────
# __vsc_ipynb_file__ is injected by VS Code with the notebook's absolute path.
# __file__ does NOT exist in Jupyter kernels, so we use None as the fallback.
_nb_str = globals().get('__vsc_ipynb_file__')  # absolute path string, or None

if _nb_str:
    # Notebook is: …/app/services/msd_to_db/explore.ipynb
    # parents[0] = msd_to_db, [1] = services, [2] = app, [3] = project root
    PROJECT_ROOT = Path(_nb_str).resolve().parents[3]
else:
    # Fallback: cwd is usually the project root when launched from VS Code terminal
    PROJECT_ROOT = Path.cwd()

FILE = PROJECT_ROOT / 'files' / 'source-db' / 'Work Order Product Advanced Find View All.xlsx'

print('Root  :', PROJECT_ROOT)
print('File  :', FILE)
print('Exists:', FILE.exists())
assert FILE.exists(), f'File not found — check PROJECT_ROOT above'
print('Size  :', f'{FILE.stat().st_size / 1_048_576:.1f} MB')
```

```python
# ── Sheet inventory ──────────────────────────────────────────────────────────
xl = pd.ExcelFile(FILE)
print("Sheets:", xl.sheet_names)
```

## 1  Full view  —  `Work Order Advanced Find View`

```python
df_full = pd.read_excel(FILE, sheet_name="Work Order Product Advanced...")

# Tidy up column names
df_full.columns = df_full.columns.str.strip()

print(f"Shape: {df_full.shape}")
print("\nColumns:")
for i, col in enumerate(df_full.columns):
    print(f"  [{i:02d}] {col}")
```

```python
df_full.head(3)
```

```python
# Data types & null counts
summary = pd.DataFrame({
    "dtype"     : df_full.dtypes,
    "non_null"  : df_full.notna().sum(),
    "null_pct"  : (df_full.isna().mean() * 100).round(1),
    "unique"    : df_full.nunique(),
})
summary
```

## 4  Quick analysis on the full view

```python
# Work order status distribution
print("=== Work Order Status ===")
print(df_full["Work Order Status"].value_counts(dropna=False))
```

```python
# Work order type distribution
print("=== Work Order Type ===")
print(df_full["Work Order Type"].value_counts(dropna=False))
```

```python
# Premier service distribution
print("=== Premier Service ===")
print(df_full["Premier Service"].value_counts(dropna=False))
```

```python
# Order type distribution
print("=== Order Type ===")
print(df_full["Order Type"].value_counts(dropna=False))
```

```python
# Date range of 'Created On'
created = pd.to_datetime(df_full["Created On"], errors="coerce")
print(f"Created On  —  min: {created.min().date()}   max: {created.max().date()}")
```

```python
# Monthly WO creation trend
df_full["created_month"] = pd.to_datetime(df_full["Created On"], errors="coerce").dt.to_period("M")
monthly = df_full.groupby("created_month").size().rename("WO_count")
print(monthly.to_string())
```

```python
# Top 10 cities by WO volume
print("=== Top 10 Cities ===")
print(df_full["City"].value_counts(dropna=False).head(10))
```

```python
# Top 10 labor vendors
print("=== Top 10 Labor Vendors ===")
print(df_full["Customer (Labor Vendor Related) (Partner Function)"].value_counts(dropna=False).head(10))
```

```python
# Closing code distribution
print("=== Closing Code ===")
print(df_full["Closing Code"].value_counts(dropna=False).head(15))
```

```python
# Repeat repair rate
rr = df_full["Repeat Repair"].value_counts(dropna=False)
print("=== Repeat Repair ===")
print(rr)
yes_count = rr.get("Yes", 0)
total = rr.sum()
print(f"\nRepeat repair rate: {yes_count / total * 100:.1f}%")
```

```python
# Top 10 product IDs (MTM)
print("=== Top 10 Product IDs (MTM) ===")
print(df_full["Product ID (MTM)"].value_counts(dropna=False).head(10))
```
